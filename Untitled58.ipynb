```python
# 1. Install required libraries
!pip install -q transformers torch

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

# 2. Define the input paragraph
original_text = (
    "Artificial intelligence is rapidly transforming industries across the globe by automating routine tasks and enhancing data-driven decision-making. "
    "In healthcare, machine learning algorithms assist doctors in diagnosing diseases with greater accuracy from medical scans. "
    "Meanwhile, the financial sector relies heavily on predictive models for fraud detection and algorithmic trading. "
    "However, this technological advancement also brings challenges, including ethical concerns surrounding data privacy and workforce displacement. "
    "Therefore, establishing robust regulatory frameworks is essential to ensure responsible AI development in the future."
)

print("="*40 + " ORIGINAL TEXT " + "="*40)
print(original_text)
print("="*95 + "\n")

# --- MODEL 1: T5-Small ---
print("Loading and running T5-Small...")
t5_name = "t5-small"
t5_tokenizer = AutoTokenizer.from_pretrained(t5_name)
t5_model = AutoModelForSeq2SeqLM.from_pretrained(t5_name)

# T5 requires a task prefix for text-to-text generation tasks
t5_inputs = t5_tokenizer("summarize: " + original_text, return_tensors="pt", max_length=512, truncation=True)
t5_summary_ids = t5_model.generate(**t5_inputs, max_length=60, min_length=20, do_sample=False)
t5_summary = t5_tokenizer.decode(t5_summary_ids[0], skip_special_tokens=True)


# --- MODEL 2: BART-Base ---
print("Loading and running BART-Base...")
bart_name = "facebook/bart-base"
bart_tokenizer = AutoTokenizer.from_pretrained(bart_name)
bart_model = AutoModelForSeq2SeqLM.from_pretrained(bart_name)

# BART does not require an explicit text prefix like T5
bart_inputs = bart_tokenizer(original_text, return_tensors="pt", max_length=512, truncation=True)
bart_summary_ids = bart_model.generate(**bart_inputs, max_length=60, min_length=20, do_sample=False)
bart_summary = bart_tokenizer.decode(bart_summary_ids[0], skip_special_tokens=True)


# --- PRINT COMPARISON ---
print("\n" + "="*40 + " T5-SMALL SUMMARY " + "="*40)
print(t5_summary)

print("\n" + "="*40 + " BART-BASE SUMMARY " + "="*40)
print(bart_summary)
```
