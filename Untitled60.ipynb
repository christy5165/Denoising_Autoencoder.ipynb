```python
# 1. Install required libraries
!pip install -q transformers torch

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# 2. Load GPT-2 Model and Tokenizer
print("Loading GPT-2 model and tokenizer...")
model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Ensure pad token is defined for generation safety
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# 3. Define the Input Prompt
prompt = "Artificial Intelligence is transforming the world because"
inputs = tokenizer(prompt, return_tensors="pt")

# 4. Generate Text using Sampling Parameters
print("Generating text...")
outputs = model.generate(
    **inputs,
    max_new_tokens=60,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    top_p=0.9,
    pad_token_id=tokenizer.eos_token_id
)

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

# 5. Display the Output
print("\n" + "="*40 + " GENERATED OUTPUT " + "="*40)
print(generated_text)
```
