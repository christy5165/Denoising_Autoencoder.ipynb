```python
# 1. Install required libraries
!pip install -q transformers torch

from transformers import pipeline

# 2. Load the pre-trained sentiment analysis pipeline
print("Loading DistilBERT sentiment model...")
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# 3. Create 3 sample tweets
tweets = [
    "I am absolutely thrilled with how well this new coding project turned out!",
    "The application keeps crashing every single time I try to save my file.",
    "It was a pretty standard day at the office, nothing too exciting."
]

print("\n" + "="*45 + " SENTIMENT CLASSIFICATION RESULTS " + "="*45)

# 4. Classify and display results
for i, tweet in enumerate(tweets, 1):
    result = sentiment_pipeline(tweet)[0]
    label = result['label']
    score = result['score']

    print(f"Tweet {i}: '{tweet}'")
    print(f"   -> Result: {label} (Confidence Score: {score:.4f})\n")
```
