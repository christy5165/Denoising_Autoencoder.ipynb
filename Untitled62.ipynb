```python
# 1. Install required libraries
!pip install -q transformers torch language-tool-python

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import language_tool_python

# 2. Load GPT-2 Model and Tokenizer
print("Loading GPT-2 model and tokenizer...")
model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# 3. Generate Text from Prompt
prompt = "The importance of clean energy is"
inputs = tokenizer(prompt, return_tensors="pt")

print("Generating text...")
outputs = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    top_p=0.9,
    pad_token_id=tokenizer.eos_token_id
)

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

# 4. Check Fluency & Grammar using the local LanguageTool instance
print("Analyzing fluency and grammar with LanguageTool...")
with language_tool_python.LanguageTool('en-US') as tool:
    matches = tool.check(generated_text)
    corrected_text = tool.correct(generated_text)

# 5. Display Results
print("\n" + "="*40 + " GENERATED TEXT " + "="*40)
print(generated_text)

print("\n" + "="*40 + " GRAMMAR & FLUENCY REPORT " + "="*40)
print(f"Total issues detected: {len(matches)}")
for i, match in enumerate(matches, 1):
    print(f"{i}. {match.message} (Context: '{match.context}')")

print("\n" + "="*40 + " SUGGESTED CORRECTION " + "="*40)
print(corrected_text)
```
