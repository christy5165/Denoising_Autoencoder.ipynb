```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Input: Sample text documents (corpus)
corpus = [
    "Artificial intelligence is rapidly transforming the modern world!",
    "Machine learning models require clean data and robust preprocessing.",
    "Natural language processing enables computers to understand human text.",
    "Artificial intelligence and machine learning drive modern innovation."
]

print("="*40 + " ORIGINAL CORPUS " + "="*40)
for i, doc in enumerate(corpus):
    print(f"Doc {i+1}: {doc}")

# 2. Initialize TfidfVectorizer
# Built-in parameters handle lowercasing, punctuation stripping, and stop word removal natively.
vectorizer = TfidfVectorizer(
    lowercase=True,          # Convert text to lowercase
    stop_words='english',    # Remove common English stop words
    token_pattern=r'(?u)\b\w\w+\b' # Automatically filters out single-character punctuation/symbols
)

# 3. Fit and Transform the text data into TF-IDF features
tfidf_matrix = vectorizer.fit_transform(corpus)

# Retrieve feature names (unique vocabulary words)
feature_names = vectorizer.get_feature_names_out()

# 4. Display the Resulting Feature Matrix as a Pandas DataFrame
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)

print("\n" + "="*40 + " TF-IDF FEATURE MATRIX " + "="*40)
pd.set_option('display.max_columns', None)
print(df_tfidf.round(4))
```
