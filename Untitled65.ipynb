```python
import re

# 1. Create a list of 5 sample tweets containing URLs, mentions, and hashtags
raw_tweets = [
    "Excited for the new AI conference! Check it out at https://example.com/ai2026 @TechLead #ArtificialIntelligence",
    "Just tried out the new #GenerativeAI model by @OpenAI. Read the review on https://t.co/xyz123 🎉",
    "Loving the sunshine today! ☀️ No coding, just relaxing. @friend1 @friend2 #weekendvibes",
    "Huge update dropped for Python libraries today. Download now at https://github.com/releases #coding #python",
    "Can someone help me debug this error? Tagging @developer_guru #help #codinglife"
]

# 2. Define a cleaning function using regular expressions
def clean_tweet(tweet):
    # Remove URLs (http/https or www links)
    tweet = re.sub(r'https?://\S+|www\.\S+', '', tweet)
    # Remove user mentions (@username)
    tweet = re.sub(r'@\w+', '', tweet)
    # Remove hashtags (#tag)
    tweet = re.sub(r'#\w+', '', tweet)
    # Clean up extra whitespace and trailing spaces left behind
    tweet = re.sub(r'\s+', ' ', tweet).strip()
    return tweet

# 3. Process and display results
print("="*55 + " TWEET PREPROCESSING PIPELINE " + "="*55)

for i, raw in enumerate(raw_tweets, 1):
    cleaned = clean_tweet(raw)
    print(f"Tweet {i}:")
    print(f"  [Original]: {raw}")
    print(f"  [Cleaned] : {cleaned}\n")
```
