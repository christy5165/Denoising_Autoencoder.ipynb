```python
# 1. Define a dictionary of keywords and their corresponding responses
keyword_responses = {
    "price": "Our pricing plans start at $9.99/month with full access to all core features.",
    "delivery": "Standard delivery usually takes 3 to 5 business days depending on your location.",
    "support": "Our customer support team is available 24/7 via live chat and email.",
    "refund": "We offer a 30-day hassle-free money-back guarantee on all initial purchases."
}

# 2. Define a list of sample user inputs to test the chatbot
user_inputs = [
    "What is the price of your service?",
    "How fast is your delivery?",
    "I need help with technical support.",
    "Can I request a refund?",
    "What's the weather like today?"  # Unmatched query
]

# 3. Create the keyword matching chatbot function
def keyword_chatbot(user_message):
    message_lower = user_message.lower()

    # Iterate through keywords to check if any exist in the user message
    for keyword, response in keyword_responses.items():
        if keyword in message_lower:
            return response

    # Default fallback if no keyword matches
    return "Sorry, I didn’t understand that."

# 4. Simulate and display responses
print("="*50 + " KEYWORD-BASED CHATBOT SIMULATION " + "="*50)

for i, text in enumerate(user_inputs, 1):
    reply = keyword_chatbot(text)
    print(f"User Input {i}: '{text}'")
    print(f"Chatbot Reply: {reply}\n")
```
