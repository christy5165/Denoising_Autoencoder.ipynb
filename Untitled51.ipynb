```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# 1. Short Text Paragraph Dataset
text = "Artificial intelligence is changing the world rapidly. Deep learning models learn patterns from data. Simple RNNs process sequential information step by step."

# 2. Character-level Preprocessing (Mapping characters to integers)
chars = sorted(list(set(text)))
char_to_idx = {ch: i for i, ch in enumerate(chars)}
idx_to_char = {i: ch for i, ch in enumerate(chars)}
vocab_size = len(chars)

print(f"Total Unique Characters (Vocab Size): {vocab_size}")

# 3. Create Input Sequences and Targets using a Sliding Window
seq_length = 20
X = []
y = []

for i in range(len(text) - seq_length):
    seq_in = text[i:i + seq_length]
    seq_out = text[i + seq_length]
    X.append([char_to_idx[char] for char in seq_in])
    y.append(char_to_idx[seq_out])

X = np.array(X)
y = np.array(y)

# 4. Build SimpleRNN Model Architecture
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=16, input_length=seq_length),
    SimpleRNN(64, return_sequences=False),
    Dense(vocab_size, activation='softmax')
])

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# 5. Train the Model
epochs = 100
print("Training the RNN model...")
history = model.fit(X, y, epochs=epochs, verbose=0)
print(f"Final Training Accuracy: {history.history['accuracy'][-1]:.4f}")

# 6. Text Generation Function
def generate_text(seed_text, length=50):
    result = seed_text
    for _ in range(length):
        # Format the seed text into the expected input length
        encoded = [char_to_idx.get(ch, 0) for ch in result[-seq_length:]]
        encoded = np.array([encoded])

        # Predict the next character probabilities
        prediction = model.predict(encoded, verbose=0)
        predicted_idx = np.argmax(prediction)

        # Append the predicted character to the result
        result += idx_to_char[predicted_idx]

    return result

# 7. Test Generation
seed = "Artificial intel"
generated_output = generate_text(seed, length=40)

print("\n--- Generation Result ---")
print(f"Seed: {seed}")
print(f"Generated: {generated_output}")
```
