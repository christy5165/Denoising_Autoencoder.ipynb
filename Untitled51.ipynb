```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Prepare a Small Dataset (Sentiment Analysis)
texts = [
    "I love this product, it is amazing!",
    "Absolute waste of money, terrible quality.",
    "Very happy with my purchase, highly recommend.",
    "Worst experience ever, broken on arrival.",
    "Fantastic service and wonderful support!",
    "Very disappointed, do not buy this."
]
# Labels: 1 for Positive, 0 for Negative
labels = np.array([1, 0, 1, 0, 1, 0])

# 2. Tokenize and Pad Sequences
vocab_size = 1000
max_length = 10
padding_type = 'post'
trunc_type = 'post'

tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(texts)

# Convert texts to integer sequences
sequences = tokenizer.texts_to_sequences(texts)

# Pad sequences so they all have the exact same length
padded = pad_sequences(sequences, maxlen=max_length, padding=padding_type, truncating=trunc_type)

# 3. Build Neural Network Architecture
model = Sequential([
    # Embedding layer maps sparse word indices into dense semantic vectors
    Embedding(input_dim=vocab_size, output_dim=16, input_length=max_length),
    # Averages word vectors across the sequence length to feed into dense layers
    GlobalAveragePooling1D(),
    Dense(16, activation='relu'),
    # Sigmoid output layer for binary classification (outputs a probability between 0 and 1)
    Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# 4. Train the Model
epochs = 25
print("Training the text classification model...")
history = model.fit(padded, labels, epochs=epochs, verbose=0)
print(f"Final Training Accuracy: {history.history['accuracy'][-1]:.4f}")

# 5. Test Inference on New Sentences
test_sentences = [
    "This is a wonderful tool, works great!",
    "Terrible product, completely broken."
]

test_sequences = tokenizer.texts_to_sequences(test_sentences)
test_padded = pad_sequences(test_sequences, maxlen=max_length, padding=padding_type, truncating=trunc_type)

predictions = model.predict(test_padded, verbose=0)

print("\n--- Predictions on New Samples ---")
for sentence, pred in zip(test_sentences, predictions):
    sentiment = "Positive" if pred[0] > 0.5 else "Negative"
    print(f"Text: '{sentence}'")
    print(f"-> Probability Score: {pred[0]:.4f} ({sentiment})\n")
```
