```python
# 1. Install graphviz package
!pip install -q graphviz

import graphviz

# 2. Initialize a directed graph
dot = graphviz.Digraph(comment='Speech-to-Summary Pipeline Workflow', format='png')
dot.attr(rankdir='TB', size='8,5')

# Configure default node styles for a polished look
dot.attr('node', fontname='Helvetica', fontsize='12')

# 3. Define Pipeline Nodes
dot.node('A', 'Audio Input\n(e.g., .mp3 / .wav file)', shape='oval', style='filled', fillcolor='#e1f5fe')
dot.node('B', 'Speech-to-Text Conversion\n(OpenAI Whisper Encoder-Decoder)', shape='box', style='filled', fillcolor='#fff9c4')
dot.node('C', 'Text Summarization\n(BART / T5 Transformer)', shape='box', style='filled', fillcolor='#fff9c4')
dot.node('D', 'Output Summary\n(Condensed Text Result)', shape='oval', style='filled', fillcolor='#e8f5e9')

# 4. Define Workflow Edges (Data Flow)
dot.edge('A', 'B', label=' Raw Audio Stream')
dot.edge('B', 'C', label=' Full Transcription')
dot.edge('C', 'D', label=' Abstractive Summary')

# 5. Display the Workflow Diagram directly inside Colab
dot
```
