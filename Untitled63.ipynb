```python
# 1. Install TextBlob library
!pip install -q textblob

from textblob import TextBlob

# 2. Create a small list of 5 sample tweets
tweets = [
    "I absolutely love the new features in this app, amazing update!",
    "The customer service was terribly slow and unhelpful today.",
    "Just grabbed a cup of coffee and started working on my project.",
    "Worst experience ever, the website crashed right at checkout.",
    "The weather today is okay, neither too hot nor too cold."
]

print("="*50 + " TWEET SENTIMENT ANALYSIS " + "="*50)

# 3. Analyze sentiment polarity for each tweet
for i, tweet in enumerate(tweets, 1):
    # Create a TextBlob object
    analysis = TextBlob(tweet)

    # Get sentiment polarity (ranges from -1.0 [most negative] to 1.0 [most positive])
    polarity = analysis.sentiment.polarity

    # Classify based on thresholds
    if polarity > 0.05:
        sentiment = "Positive"
    elif polarity < -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    # 4. Print the tweet, score, and classification
    print(f"Tweet {i}: {tweet}")
    print(f"   -> Polarity Score: {polarity:.2f} | Sentiment: {sentiment}\n")
```
