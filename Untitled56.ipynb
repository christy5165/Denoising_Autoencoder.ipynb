```python
# 1. Install required libraries
!pip install -q transformers torch matplotlib seaborn

import torch
import matplotlib.pyplot as plt
import seaborn as sns
from transformers import T5Config, T5Tokenizer, T5ForConditionalGeneration

# 2. Configure model to explicitly output attentions from the start
print("Loading t5-small with attention configuration...")
model_name = "t5-small"
config = T5Config.from_pretrained(model_name, output_attentions=True)
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name, config=config)

# 3. Define input text with the required T5 summarization prefix
text = "Artificial intelligence is rapidly transforming global industries by automating routine tasks."
input_text = "summarize: " + text

# Tokenize inputs
inputs = tokenizer(input_text, return_tensors="pt")

# 4. Run forward pass on the encoder directly
encoder = model.get_encoder()
with torch.no_grad():
    encoder_outputs = encoder(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"]
    )

# Safely check and extract attentions tuple
encoder_attentions = encoder_outputs.attentions

if encoder_attentions is not None and len(encoder_attentions) > 0:
    layer_idx = 0
    head_idx = 0
    # Shape: (batch_size, num_heads, seq_len, seq_len)
    attn_matrix = encoder_attentions[layer_idx][0, head_idx].detach().numpy()

    # Convert token IDs to readable string tokens
    tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

    # 5. Plotting the Attention Matrix Heatmap
    plt.figure(figsize=(10, 8))
    sns.heatmap(
        attn_matrix,
        xticklabels=tokens,
        yticklabels=tokens,
        cmap="Blues",
        cbar=True
    )
    plt.title(f"T5-Small Encoder Self-Attention (Layer {layer_idx}, Head {head_idx})", fontsize=14)
    plt.xlabel("Key Tokens", fontsize=12)
    plt.ylabel("Query Tokens", fontsize=12)
    plt.xticks(rotation=90)
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
else:
    print("Attention weights were not returned. Please check model config settings.")
```
