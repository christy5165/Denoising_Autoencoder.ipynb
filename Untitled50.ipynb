```python
# 1. Install and download required NLTK data packages
!pip install -q nltk

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download necessary NLTK resource files
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

# 2. Create raw text input
raw_text = (
    "Natural Language Processing is an exciting field of Artificial Intelligence! "
    "It enables computers to understand, interpret, and manipulate human language. "
    "Are you ready to build amazing GenAI apps?"
)

print("="*40 + " RAW TEXT " + "="*40)
print(raw_text)

# 3. Convert text to lowercase
lowercased_text = raw_text.lower()

# 4. Tokenize the text into individual words
tokens = word_tokenize(lowercased_text)

# 5. Remove punctuation and stop words
stop_words = set(stopwords.words('english'))

# Filter tokens: keep only alphanumeric words that are not in the stop words list
cleaned_tokens = [
    word for word in tokens
    if word.isalnum() and word not in stop_words
]

# 6. Display Results
print("\n" + "="*40 + " TOKENIZED TOKENS " + "="*40)
print(tokens)

print("\n" + "="*40 + " CLEANED & FILTERED TOKENS " + "="*40)
print(cleaned_tokens)
```
