```python
# 1. Install TextBlob library
!pip install -q textblob

from textblob import TextBlob

# 2. Define a list of sample user inputs to test the emotion chatbot
user_inputs = [
    "I am so happy with how fast my issue was resolved!",
    "The service was terrible and I am very frustrated right now.",
    "Today is Tuesday and I am checking my account balance."
]

# 3. Create the emotion-aware chatbot function
def emotion_chatbot(user_message):
    # Analyze sentiment polarity using TextBlob (-1.0 to 1.0)
    blob = TextBlob(user_message)
    polarity = blob.sentiment.polarity

    # Branch responses based on sentiment thresholds
    if polarity > 0.05:
        return "I’m glad to hear that!"
    elif polarity < -0.05:
        return "I’m sorry to hear that. How can I help?"
    else:
        return "Thanks for sharing."

# 4. Simulate and display responses
print("="*50 + " EMOTION-AWARE CHATBOT SIMULATION " + "="*50)

for i, text in enumerate(user_inputs, 1):
    reply = emotion_chatbot(text)
    print(f"User Input {i}: '{text}'")
    print(f"Chatbot Reply: {reply}\n")
```
