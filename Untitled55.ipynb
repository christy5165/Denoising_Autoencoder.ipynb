```python
# 1. Install required libraries and system audio dependencies
!pip install -q openai-whisper transformers torch
!apt-get install -qq ffmpeg

import whisper
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import os

# 2. Load Whisper Model for Transcription
print("Loading Whisper speech recognition model...")
whisper_model = whisper.load_model("base")

# 3. Load Summarization Model (BART-base)
print("Loading summarization model...")
summarizer_name = "facebook/bart-base"
tokenizer = AutoTokenizer.from_pretrained(summarizer_name)
summarizer_model = AutoModelForSeq2SeqLM.from_pretrained(summarizer_name)

# 4. Specify Audio File Path
# (Upload any short spoken audio file via the Colab sidebar, e.g., 'speech_sample.mp3')
audio_filename = "speech_sample.mp3"

if not os.path.exists(audio_filename):
    print(f"\n[Note]: '{audio_filename}' was not found.")
    print("Please upload a short audio file using the left-side file panel in Colab before running.")
else:
    # --- STEP A: Speech Recognition (Audio to Text) ---
    print(f"\nTranscribing audio file: {audio_filename}...")
    transcription_result = whisper_model.transcribe(audio_filename)
    transcribed_text = transcription_result["text"].strip()

    # --- STEP B: Text Summarization (Text to Summary) ---
    print("Generating summary from transcription...")
    inputs = tokenizer(transcribed_text, return_tensors="pt", max_length=512, truncation=True)

    summary_ids = summarizer_model.generate(
        **inputs,
        max_length=50,
        min_length=15,
        do_sample=False
    )
    summarized_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

    # --- STEP C: Display Results ---
    print("\n" + "="*40 + " TRANSCRIPTION (SPEECH-TO-TEXT) " + "="*40)
    print(transcribed_text)

    print("\n" + "="*40 + " GENERATED SUMMARY " + "="*40)
    print(summarized_text)
```
