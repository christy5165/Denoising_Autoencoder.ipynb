```python
# 1. Install OpenAI Whisper and system audio dependency (ffmpeg)
!pip install -q openai-whisper
!apt-get install -qq ffmpeg

import whisper
import os

# 2. Load the Whisper model (using the 'base' model for a good balance of speed and accuracy)
print("Loading Whisper model...")
model = whisper.load_model("base")

# 3. Specify your audio file path
# (You can upload an audio file via the Colab sidebar and update its path here, e.g., 'sample.mp3')
audio_filename = "sample.mp3"

if not os.path.exists(audio_filename):
    print(f"\n[Note]: '{audio_filename}' was not found in the current directory.")
    print("Please upload a short audio file (<10 seconds) using the left-side file panel in Colab.")
else:
    # 4. Transcribe the audio file
    print(f"Transcribing audio file: {audio_filename}...")
    result = model.transcribe(audio_filename)
    transcribed_text = result["text"]

    # 5. Display the Results
    print("\n" + "="*40 + " TRANSCRIPTION RESULT " + "="*40)
    print(f"Audio File Name: {audio_filename}")
    print(f"Transcribed Text: {transcribed_text.strip()}")
```
