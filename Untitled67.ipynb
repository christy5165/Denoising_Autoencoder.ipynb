```python
# 1. Define a list of sample customer service queries
user_queries = [
    "What are your working hours?",
    "How can I reset my password?",
    "Where is my order?",
    "Do you offer refunds?",
    "Can I speak to a live human agent?"
]

# 2. Define a rule-based chatbot function using if-elif-else conditions
def customer_service_bot(query):
    # Convert query to lowercase to ensure case-insensitive matching
    q = query.lower()

    if "working hours" in q or "hours" in q:
        return "Our support team is available Monday through Friday, 9:00 AM to 6:00 PM."
    elif "password" in q or "reset" in q:
        return "You can reset your password by clicking 'Forgot Password' on the login screen and following the email link."
    elif "order" in q or "shipping" in q:
        return "You can track your recent orders by visiting the 'My Orders' section in your account dashboard."
    elif "refund" in q or "return" in q:
        return "We offer a 30-day hassle-free return policy on all unused items. Visit our returns portal to begin."
    else:
        return "I'm sorry, I didn't quite understand your request. Please email support@example.com for further assistance."

# 3. Test the chatbot across all sample queries
print("="*55 + " CUSTOMER SERVICE CHATBOT SIMULATION " + "="*55)

for i, query in enumerate(user_queries, 1):
    reply = customer_service_bot(query)
    print(f"User Query {i}: '{query}'")
    print(f"Chatbot Reply: '{reply}'\n")
```
